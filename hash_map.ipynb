```python
# find indices of two number that add up to target

nums = [2, 7, 11, 15]
target = 9

mp = {}

for i in range(len(nums)):
    complement = target - nums[i]

    if complement in mp:
        print([mp[complement], i])
        break

    mp[nums[i]] = i


```
